```python
user_counts = final_rating['user_id'].value_counts()
filtered_users = user_counts[user_counts > 200].index
```

```python
book_counts = final_rating['title'].value_counts()
filtered_books = book_counts[book_counts > 100].index
```

```python
filtered_data = final_rating[
    (final_rating['user_id'].isin(filtered_users)) &
    (final_rating['title'].isin(filtered_books))
]
```

```python
book_pivot = filtered_data.pivot_table(index='title', columns='user_id', values='rating_x').fillna(0)
```

```python
from sklearn.neighbors import NearestNeighbors # type: ignore
model = NearestNeighbors(metric='cosine',algorithm='brute')
```

```python
model.fit(book_pivot.values)
```

```python
distance, suggestion = model.kneighbors(book_pivot.iloc[276,:].values.reshape(1,-1),n_neighbors = 6)
```

```python
def recommend_book(book_name):
    book_id = np.where(book_pivot.index == book_name)[0][0]
    distance, suggestion = model.kneighbors(book_pivot.iloc[book_id,:].values.reshape(1,-1),n_neighbors = 6)

    for i in range(len(suggestion)):
      books = book_pivot.index[suggestion[i]]  
    for j in books:
        print(j)
```

```python
book_name = 'Harry Potter and the Chamber of Secrets (Book 2)'
recommend_book(book_name)
```
